```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Sample data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Create line plot
sns.lineplot(x=x, y=y)
plt.title('Basic Line Plot')
plt.show()
```

```python
categories = ['A', 'B', 'C', 'D']
values = [10, 20, 15, 25]

sns.barplot(x=categories, y=values)
plt.title('Bar Plot Example')
plt.show()
```

```python
import pandas as pd

# Create DataFrame
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May'],
    'Sales': [120, 150, 180, 90, 110],
    'Expenses': [80, 90, 100, 70, 85]
}
df = pd.DataFrame(data)

# Line plot from DataFrame
sns.lineplot(data=df, x='Month', y='Sales')
plt.title('Monthly Sales')
plt.show()
```

```python
tips = sns.load_dataset('tips')  # Built-in dataset

sns.scatterplot(data=tips, x='total_bill', y='tip')
plt.title('Total Bill vs Tip')
plt.show()
```

```python
sns.regplot(data=tips, x='total_bill', y='tip')
plt.title('Regression Plot')
plt.show()
```

```python
iris = sns.load_dataset('iris')
sns.pairplot(iris, hue='species')
plt.show()
```

```python
sns.kdeplot(data=tips, x='total_bill', hue='time', fill=True)
plt.title('KDE of Total Bill by Meal Time')
plt.show()
```

```python
sns.ecdfplot(data=tips, x='total_bill', hue='time')
plt.title('ECDF of Total Bill by Meal Time')
plt.show()
```

```python
sns.clustermap(iris.drop('species', axis=1).corr(), annot=True)
plt.show()
```

```python
g = sns.FacetGrid(tips, col='time', row='smoker')
g.map(sns.scatterplot, 'total_bill', 'tip')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

sns.scatterplot(data=tips, x='total_bill', y='tip', ax=ax1)
sns.boxplot(data=tips, x='day', y='total_bill', ax=ax2)

plt.tight_layout()
plt.show()
```

```python
sns.lmplot(data=tips, x='total_bill', y='tip', hue='smoker', 
           col='time', row='sex', height=4)
plt.show()
```
